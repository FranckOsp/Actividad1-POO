```python
#EJERCICIO RESUELTO 4

class Edades:

    @staticmethod
    def calcular_edalber(edjuan: float) -> float:
        return 2 * edjuan / 3

    @staticmethod
    def calcular_edana(edjuan: float) -> float:
        return 4 * edjuan / 3

    @staticmethod
    def calcular_edmama(edjuan: float, edalber: float, edana: float) -> float:
        return edjuan + edalber + edana


def main():
    #Ejercicio 4
    edjuan = float(input("Cuantos años tiene Juan? "))

    edalber = Edades.calcular_edalber(edjuan)
    edana = Edades.calcular_edana(edjuan)
    edmama = Edades.calcular_edmama(edjuan, edalber, edana)

    print(f"la edad de la mama es: {edmama}")
    print(f"la edad de juan es: {edjuan}")
    print(f"la edad de alber es: {edalber}")
    print(f"la edad de ana es: {edana}")

if __name__ == "__main__":
    main()
```

```python
#EJERCICIO RESUELTO 5

import math

class CalculoOperaciones:

    @staticmethod
    def calcular_suma(suma: float, x: float, y: float) -> float:
        suma = suma + x
        x = x + (y ** 2)
        suma = suma + (x / y)
        return suma


def main():
    print("--- INGRESO DE DATOS ---")
    suma_inicial = float(input("Ingrese el valor inicial de SUMA: "))
    x_inicial = float(input("Ingrese el valor inicial de X: "))
    y_inicial = float(input("Ingrese el valor inicial de Y: "))
    resultado_final = CalculoOperaciones.calcular_suma(suma_inicial, x_inicial, y_inicial)

    print(f"\nEL VALOR DE LA SUMA ES: {resultado_final}")


if __name__ == "__main__":
    main()
```

```python
#EJERCICIO RESUELTO 12

class SalarioEmpleado:

    @staticmethod
    def calcular_salario_bruto(horas: float, valor_hora: float) -> float:
        return horas * valor_hora

    @staticmethod
    def calcular_retencion(salario_bruto: float, porcentaje_retencion: float) -> float:
        return salario_bruto * (porcentaje_retencion / 100)

    @staticmethod
    def calcular_salario_neto(salario_bruto: float, retencion: float) -> float:
        return salario_bruto - retencion


def main():
    horas = 48.0
    valor_hora = 5000.0
    porcentaje_retencion = 12.5
    salario_bruto = SalarioEmpleado.calcular_salario_bruto(horas, valor_hora)
    retencion = SalarioEmpleado.calcular_retencion(salario_bruto, porcentaje_retencion)
    salario_neto = SalarioEmpleado.calcular_salario_neto(salario_bruto, retencion)
    print("=== LIQUIDACIÓN DEL EMPLEADO ===")
    print(f"Salario Bruto: ${salario_bruto:,.2f}")
    print(f"Retención en la Fuente ({porcentaje_retencion}%): ${retencion:,.2f}")
    print(f"Salario Neto: ${salario_neto:,.2f}")


if __name__ == "__main__":
    main()
```

```python
#EJERCICIO RESUELTO 14

class Calculos:

  @staticmethod
  def calcular_cuadrado(numero: float) -> float:
    return numero ** 2

  @staticmethod
  def calcular_cubo(numero: float) -> float:
    return numero ** 3

def main():
  x = float(input((f"Ingrese el valor a calcular: ")))
  cuadrado_x = Calculos.calcular_cuadrado(x)
  cubo_x = Calculos.calcular_cubo(x)
  print("==CALCULO DE CUADRADO Y CUBO==")
  print(f"El cuadrado de {x} es {cuadrado_x:,.2f}")
  print(f"El cubo de {x} es {cubo_x:,.2f}")

if __name__ == "__main__":
  main()
```

```python
#EJERCICIO RESUELTO 17

class OperacionesCirculo:

  @staticmethod
  def Calcular_area(radio: float) -> float:
    return math.pi * (radio ** 2)

  @staticmethod
  def Calcular_circunferencia(radio: float) -> float:
    return 2 * math.pi * radio

def main():
  radio = float(input(f"Ingrese el valor del radio del circulo: "))
  area = OperacionesCirculo.Calcular_area(radio)
  circunferencia = OperacionesCirculo.Calcular_circunferencia(radio)
  print("==RESULTADOS==")
  print(f"El Area del circulo es {area:,.2f}")
  print(f"La Circunferencia del circulo es {circunferencia:,.2f}")

if __name__ == "__main__":
  main()
```
